```python
import pandas as pd
```

```python
praise_df = pd.read_csv(
    "../data/praise_dataset.csv",
    header=None,
    names=["content"]
)

complaint_df = pd.read_csv(
    "../data/complaint.csv",
    header=None,
    names=["content"]
)

bug_df = pd.read_csv(
    "../data/bug_report_df.csv",
    header=None,
    names=["content"]
)

question_df = pd.read_csv(
    "../data/questions.csv",
    header=None,
    names=["content"]
)

suggestion_df = pd.read_csv(
    "../data/suggestion_df.csv",
    header=None,
    names=["content"]
)
```

```python
praise_df["category"] = "Praise"
complaint_df["category"] = "Complaint"
bug_df["category"] = "Bug Report"
question_df["category"] = "Question"
suggestion_df["category"] = "Suggestion"
```

```python
print(praise_df.columns)
print(complaint_df.columns)
print(bug_df.columns)
print(question_df.columns)
print(suggestion_df.columns)
```

```python
df = pd.concat(
    [
        praise_df,
        complaint_df,
        bug_df,
        question_df,
        suggestion_df,
    ],
    ignore_index=True
)
```
