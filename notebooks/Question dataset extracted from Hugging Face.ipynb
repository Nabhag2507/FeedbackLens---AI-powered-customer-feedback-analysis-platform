```python
import requests

url = 'https://huggingface.co/datasets/bitext/Bitext-customer-support-llm-chatbot-training-dataset/embed/viewer/default/train'

response = requests.get(url)
```

```python
df = pd.read_csv("hf://datasets/bitext/Bitext-customer-support-llm-chatbot-training-dataset/Bitext_Sample_Customer_Support_Training_Dataset_27K_responses-v11.csv")
```

```python
df[['instruction']].sample(10)
```

```python
question_words = r"\b(what|where|when|who|whose|which|why|how)\b"

question_df = df[
    df["instruction"].str.contains(
        question_words,
        case=False,
        na=False,
        regex=True
    )
]

question_df[["instruction"]].sample(10)
```

```python
final_df = question_df[["instruction"]].sample(1000, random_state = 42)
```
